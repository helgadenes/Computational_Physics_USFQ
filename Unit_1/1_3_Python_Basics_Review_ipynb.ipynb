### 7.3 Tuples:
To define tuples we use ().

Tuples -> sets of #'s

```python
# Define a tuple

tuple_1 = (100, 200, 300)

print(type(tuple_1))
```

```python
# Use indexing

print(tuple_1[0:2])
```

```python
print(len(tuple_1))
```

```python
print(tuple_1.count(100))
```

```python
# Differences between tuples and lists

list_1 = [100, 200, 300]

# Change an element within the list

list_1[2] = 500

print(list_1)

# Can we do the same with tuples? NO.

#tuple_1[2] = 500 # This gives us an error
# TypeError: 'tuple' object does not support item assignment
```

```python
# Tuples can receive objects via unpacking
# Let's get the elements from list_1

print(list_1)

x, y, z = list_1

print(x, y, z)
```

```python
# Parethesis are not strictly needed to generate tuples
# But they are highly desired

tuple_2 = 7,8,9

tuple_3 = (7,8,9)

print(type(tuple_2))
print(type(tuple_3))
```

### 7.4 Sets:
To define sets we use {} (curly brackets).
Allow for some basic operations:

Unions
Intersections
Differences

```python
{1, 1, 2, 3, 4, 4, 5, 6, 6, 7, 7, 7}
```

```python
print(type({1, 1, 2, 3, 4, 4, 5, 6, 6, 7, 7, 7}))
```

```python
# Operations via set():

# The unique elements of a list can be found like this:

set_1 = set([1, 1, 2, 3, 4, 4, 5, 6, 6, 7, 7, 7])

print(set_1)
```

```python
# unique elements for tuples

set_2 = set((1, 1, 2, 3, 4, 4, 5, 6, 6, 7, 7, 7))

print(set_2)
```

```python
# For strings:

set_3 = set("háromszög")

print(set_3)
```

```python
# Define a new set:

set_4 = set((5, 6, 7, 8, 9, 10))

print(set_4)
```

```python
# Union operation:

set_2.union(set_4)
```

```python
# Intersection:

set_2.intersection(set_4)
```

## 7.5 Dictionaries:

To define dictionaries we also use {} (curly brackets), but need to define "keys" to individual elements. Allow for some basic operations:

```python
# Define a dictionary:

dict_1 = {"CP1": 7 , "QM1": 6 , "ED1": 6, "GR": 9}

print(type(dict_1))
```

```python
# Indexing works with the keys:

print(dict_1["GR"])
```

```python
# Print out the keys available

dict_1.keys()
```

```python
# Print the length of the dictionary

print(len(dict_1))
```

```python
# Empty dictionaries:

universities_dict = {}
```

```python
# Fill up the dictionary:

universities_dict["USFQ"] = "Pichincha"

print(universities_dict)
```

```python
universities_dict["EPN"] = "Pichincha"

print(universities_dict)
```

```python
universities_dict["Espol"] = "Guayas"

print(universities_dict)
```

```python
universities_dict["Yachay"] = "Imbabura"

print(universities_dict)
```

```python
universities_dict["Ecuador code"] = 593

print(universities_dict)
```

```python
# Adding multiple elements in one go:
# https://www.programiz.com/python-programming/methods/dictionary/update

universities_dict.update({"UTA": "Tungurahua", "ESPOCH": "Chimborazo"})
```

```python
print(universities_dict)
```

```python
# Remove elements from dictionary via del()

del universities_dict["Ecuador code"]
```

```python
print(universities_dict)
```
